```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name  # Nama Hero
        self.hp = hp  # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero1.hp = 500
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()
```

```python
from abc import ABC, abstractmethod

# 1. ABSTRACTION (Kerangka Dasar)
class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar):
        self.nama_kamar = nama_kamar
        # Encapsulation: Private Attributes (menggunakan __)
        self.__harga_dasar = harga_dasar
        self.__stok = 0

    # Getter untuk stok
    def get_stok(self):
        return self.__stok

    # Setter/Method update stok dengan validasi tidak boleh negatif
    def set_stok(self, jumlah, nama_label=None):
        label = nama_label if nama_label else self.nama_kamar
        if jumlah < 0:
            print(f"Gagal update stok {label}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok = jumlah
            print(f"Berhasil menambahkan stok {label}: {jumlah} unit.")

    # Getter untuk harga dasar (agar dapat diakses aman oleh class anak)
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Abstract Method (Wajib di-override oleh class anak)
    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

    @abstractmethod
    def tampilkan_detail(self, jumlah_malam, nomor=1):
        pass



# 2. INHERITANCE & POLYMORPHISM
class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        self.fasilitas = fasilitas

    # Pajak sewa Deluxe: 10%
    def hitung_pajak(self):
        return 0.10 * self.get_harga_dasar()

    # Polymorphism: Override hitung_harga_total
    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
        return int(harga_per_malam * jumlah_malam)

    # Polymorphism: Override tampilkan_detail
    def tampilkan_detail(self, jumlah_malam, nomor=1):
        harga_dasar = self.get_harga_dasar()
        pajak = self.hitung_pajak()
        subtotal = self.hitung_harga_total(jumlah_malam)

        # Format Rupiah
        harga_dasar_str = f"Rp {int(harga_dasar):,}".replace(",", ".")
        pajak_str = f"Rp {int(pajak):,}".replace(",", ".")
        subtotal_str = f"Rp {int(subtotal):,}".replace(",", ".")

        print(f"{nomor}. [DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: {harga_dasar_str} | Pajak(10%): {pajak_str}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {subtotal_str}")


class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        self.kapasitas = kapasitas

    # Pajak sewa Standard: 5%
    def hitung_pajak(self):
        return 0.05 * self.get_harga_dasar()

    # Polymorphism: Override hitung_harga_total
    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
        return int(harga_per_malam * jumlah_malam)

    # Polymorphism: Override tampilkan_detail
    def tampilkan_detail(self, jumlah_malam, nomor=1):
        harga_dasar = self.get_harga_dasar()
        pajak = self.hitung_pajak()
        subtotal = self.hitung_harga_total(jumlah_malam)

        # Format Rupiah
        harga_dasar_str = f"Rp {int(harga_dasar):,}".replace(",", ".")
        pajak_str = f"Rp {int(pajak):,}".replace(",", ".")
        subtotal_str = f"Rp {int(subtotal):,}".replace(",", ".")

        print(f"{nomor}. [STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: {harga_dasar_str} | Pajak(5%): {pajak_str}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {subtotal_str}")



# 3. FITUR PEMESANAN (POLYMORPHISM)
def proses_transaksi(daftar_pesanan):
    print("--- STRUK PEMESANAN ---")
    total_tagihan = 0

    for idx, (kamar, jumlah_malam) in enumerate(daftar_pesanan, start=1):
        kamar.tampilkan_detail(jumlah_malam, nomor=idx)
        total_tagihan += kamar.hitung_harga_total(jumlah_malam)

    total_str = f"Rp {int(total_tagihan):,}".replace(",", ".")
    print("-" * 40)
    print(f"TOTAL TAGIHAN: {total_str}")
    print("-" * 40)



# 4. ALUR PROGRAM (USER STORY)
if __name__ == "__main__":
    print("--- SETUP DATA KAMAR ---")

    # a) Admin membuat data kamar (1 Deluxe, 1 Standard)
    kamar_deluxe = KamarDeluxe("Kamar Deluxe Sea View", 1500000, "Private Pool")
    kamar_standard = KamarStandard("Kamar Standard Superior", 500000, "2 Orang")

    # b) Admin mencoba mengisi stok kamar (termasuk pengujian angka negatif)
    kamar_deluxe.set_stok(10, nama_label="Kamar Deluxe")
    kamar_standard.set_stok(-5, nama_label="Kamar Standard")
    kamar_standard.set_stok(20, nama_label="Kamar Standard")

    # c) Tamu memesan 2 malam Kamar Deluxe dan 1 malam Kamar Standard
    daftar_pesanan = [
        (kamar_deluxe, 2),
        (kamar_standard, 1)
    ]

    # d) Menampilkan detail pemesanan dan total tagihan
    proses_transaksi(daftar_pesanan)
```

analisi 5 = masi berjalan,Polimorfisme memudahkan kita untuk menambahkan karakter baru tanpa mengubah kode yang sudah ada.

-bagian hero archer menjadi tangan kosong bukan memanah dari jauh lagi,Karena Polimorfisme menggunakan nama method yang sama sebagai perintah umum. Looping memanggil serang(), jadi semua Child Class harus memiliki method serang().

```python
from abc import ABC, abstractmethod


class KamarHotel(ABC):

    def __init__(self, nama_kamar, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = 0
        self.__harga_dasar = harga_dasar

    # Getter untuk melihat stok
    def get_stok(self):
        return self.__stok

    # Getter untuk melihat harga dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Method untuk menambah stok
    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(
                f"Gagal update stok {self.nama_kamar}! "
                f"Stok tidak boleh negatif ({jumlah})."
            )
        else:
            self.__stok += jumlah
            print(
                f"Berhasil menambahkan stok {self.nama_kamar}: "
                f"{jumlah} unit."
            )

    # Method abstrak
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        self.fasilitas = fasilitas

    # POLYMORPHISM
    def tampilkan_detail(self):
        print(
            f"[DELUXE] {self.nama_kamar} | "
            f"Fasilitas: {self.fasilitas}"
        )

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 10 / 100
        subtotal = (harga + pajak) * jumlah_malam

        print(
            f"Harga Dasar/Malam: Rp {harga:,.0f} | "
            f"Pajak(10%): Rp {pajak:,.0f}"
        )
        print(
            f"Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {subtotal:,.0f}"
        )

        return subtotal

class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        self.kapasitas = kapasitas

    # POLYMORPHISM
    def tampilkan_detail(self):
        print(
            f"[STANDARD] {self.nama_kamar} | "
            f"Kapasitas: {self.kapasitas}"
        )

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 5 / 100
        subtotal = (harga + pajak) * jumlah_malam

        print(
            f"Harga Dasar/Malam: Rp {harga:,.0f} | "
            f"Pajak(5%): Rp {pajak:,.0f}"
        )
        print(
            f"Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {subtotal:,.0f}"
        )

        return subtotal

def proses_transaksi(daftar_pesanan):

    total_tagihan = 0

    print("\n--- STRUK PEMESANAN ---")

    for nomor, pesanan in enumerate(daftar_pesanan, start=1):

        kamar = pesanan["kamar"]
        jumlah_malam = pesanan["malam"]

        print(f"{nomor}. ", end="")
        kamar.tampilkan_detail()

        subtotal = kamar.hitung_harga_total(jumlah_malam)

        total_tagihan += subtotal
        print()

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")


print("--- SETUP DATA KAMAR ---")

deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    1_500_000,
    "Private Pool"
)

standard = KamarStandard(
    "Kamar Standard Superior",
    500_000,
    "2 Orang"
)


# Menambahkan stok
deluxe.tambah_stok(10)

# Mencoba stok negatif
standard.tambah_stok(-5)

# Menambahkan stok yang benar
standard.tambah_stok(20)


daftar_pesanan = [
    {
        "kamar": deluxe,
        "malam": 2
    },
    {
        "kamar": standard,
        "malam": 1
    }
]

proses_transaksi(daftar_pesanan)
```
